```python
from paddleocr import PaddleOCR
import time

ocr_v6 = PaddleOCR(
    use_textline_orientation=True,
    lang='en',
    ocr_version='PP-OCRv6',
    enable_mkldnn=False
)
```

```python
IMAGE_PATH = r"C:\Users\ST\Downloads\q8.jpg"

start = time.time()
result_v6 = list(ocr_v6.predict(IMAGE_PATH))
end = time.time()
inference_time_v6 = end - start

predicted_lines_v6 = result_v6[0]['rec_texts']

print("Predicted lines:", predicted_lines_v6)
print("Inference time:", inference_time_v6, "seconds")
```

```python
import jiwer

ground_truth_lines = [
    "Name: Arooj Fatima",
    "Roll Number: BSDSF23M038",
    "Artificial Intelligence and OCR are transforming education.",
    "The quick brown fox jumps over the lazy dog.",
    "Python, NumPy, PyTorch and OpenCV are popular libraries. ML models require training and evaluation."
]

ref_text = " ".join(ground_truth_lines)
hyp_text_v6 = " ".join(predicted_lines_v6)

cer_v6 = jiwer.cer(ref_text, hyp_text_v6)
wer_v6 = jiwer.wer(ref_text, hyp_text_v6)
accuracy_v6 = (1 - cer_v6) * 100

print("="*50)
print("PADDLEOCR (PP-OCRv6) BENCHMARK RESULTS")
print("="*50)
print(f"CER: {cer_v6:.3f} ({cer_v6*100:.1f}%)")
print(f"WER: {wer_v6:.3f} ({wer_v6*100:.1f}%)")
print(f"Accuracy: {accuracy_v6:.2f}%")
print(f"Inference Time: {inference_time_v6:.2f} seconds")
print("="*50)
```

```python
import os

def folder_size_mb(path):
    if not os.path.exists(path):
        return None
    total = sum(
        os.path.getsize(os.path.join(dirpath, f))
        for dirpath, _, filenames in os.walk(path)
        for f in filenames
    )
    return total / (1024 ** 2)

paths = {
    "PaddleX model cache": os.path.expanduser(r"~\.paddlex\official_models"),
    "Old PaddleOCR cache": os.path.expanduser(r"~\.paddleocr"),
}

for label, p in paths.items():
    size = folder_size_mb(p)
    print(f"{label}: {size:.1f} MB" if size is not None else f"{label}: not found")
```
